```python
import os
import csv
import pandas as pd
import matplotlib.pyplot as plt

ip=[]
mydict={}

#Taking data from the log file

with open("C:\\Users\\Ashok S\\Desktop\\mylog.log")as f:

    for each_line in f:
        mylist=each_line.split(" ")
        ip.append(mylist[3].split(":")[0])


def frequency(mylist):
    for each in mylist:
        if each not in mydict:
            mydict[each]=1
        else:
            mydict[each]+=1
            
    return mydict

result=frequency(ip)

#converting to csv file

def write_to_csv(mydict):
    with open("C:\\Users\\Ashok S\\Desktop\\new_log_csv.csv","w",newline="") as f:
        writer=csv.writer(f)
        writer.writerow(["IP Address","Frequency"])
        for each in mydict:
            writer.writerow([each,mydict[each]])
write_to_csv(result)
```

```python
#Matplotlib 
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("C:\\Users\\Ashok S\\Desktop\\new_log_csv.csv")
plt.bar(df["IP Address"],df["Frequency"],edgecolor="black")
plt.title("IP ADDRESS AND NUMBER OF REQUESTS (FREQUENCY)")
plt.xlabel("IP ADDRESS")
plt.ylabel("NUMBER OF REQUESTS")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("log_project_image_1")
plt.show();
```
